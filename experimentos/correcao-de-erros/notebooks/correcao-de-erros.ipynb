```python
plt.figure(figsize=(9, 5))
ax = sns.barplot(
    data=df,
    x="tecnica",
    y="likert",
    hue="llm",
    order=order_tecnica,
    hue_order=order_llm,
    estimator="mean",
    errorbar=None,
    palette=["#4C78A8", "#F58518"]
)

ax.set_title("Media Likert na verificacao da correcao de erros", weight="bold")
ax.set_xlabel("Tecnica")
ax.set_ylabel("Media Likert")
ax.set_ylim(0, 5.2)
ax.legend(title="LLM")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4), sharey=True)

for ax, llm in zip(axes, order_llm):
    pivot = (
        df[df["llm"] == llm]
        .pivot_table(index="grafico", columns="tecnica", values="acerto", aggfunc="mean")
        .reindex(index=order_grafico, columns=order_tecnica)
    )

    sns.heatmap(
        pivot,
        ax=ax,
        annot=True,
        fmt=".0f",
        cmap="Blues",
        vmin=0,
        vmax=1,
        cbar=(llm == "Claude"),
        linewidths=0.5,
        linecolor="white"
    )
    ax.set_title(f"Acerto por item - {llm}", weight="bold")
    ax.set_xlabel("Tecnica")
    ax.set_ylabel("Grafico")

plt.suptitle("Mapa de acerto por item na correcao de erros", y=1.05, fontsize=14, weight="bold")
plt.tight_layout()
plt.show()
```
